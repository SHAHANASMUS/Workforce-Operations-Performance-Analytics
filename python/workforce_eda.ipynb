# Workforce & Operations Performance Analytics — Python EDA

Exploratory data analysis for workforce performance, engagement, productivity and business outcomes.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
pd.set_option('display.max_columns', None)
```

## Load cleaned datasets

```python
base = '../data/'
employees = pd.read_csv(base + 'employees_clean.csv')
stores = pd.read_csv(base + 'stores_clean.csv')
monthly_performance = pd.read_csv(base + 'monthly_performance_clean.csv')
role_kpis = pd.read_csv(base + 'role_kpis_clean.csv')
business_outcomes = pd.read_csv(base + 'business_outcomes_clean.csv')
print(employees.shape, stores.shape, monthly_performance.shape, role_kpis.shape, business_outcomes.shape)
```

## Workforce distribution

```python
dept_counts = employees.groupby('Department')['Employee_ID'].nunique().sort_values(ascending=False)
display(dept_counts.to_frame('Employee_Count'))
```

## Performance by department

```python
perf = monthly_performance.merge(employees[['Employee_ID','Department']], on='Employee_ID', how='left')
perf_dept = perf.groupby('Department')[['Performance_Rating','Employee_Satisfaction','Engagement_Index']].mean().sort_values('Performance_Rating', ascending=False)
display(perf_dept.round(2))
```

## Correlation analysis

```python
corr_cols = ['Manager_Evaluation','Performance_Rating','Employee_Satisfaction','Engagement_Index','Monthly_Bonus','Benefits_Cost','Absenteeism_Days']
display(monthly_performance[corr_cols].corr().round(2))
```

## Active vs exited employee profile

```python
employees['Employee_Status'] = np.where(employees['Exit_Date'].isna(), 'Active', 'Exited')
status = monthly_performance.merge(employees[['Employee_ID','Employee_Status']], on='Employee_ID', how='left')
display(status.groupby('Employee_Status')[['Performance_Rating','Employee_Satisfaction','Engagement_Index','Overtime_Hours']].mean().round(2))
```

## Business outcomes

```python
summary = business_outcomes.groupby('Department').agg(Sales_Actual=('Sales_Actual','sum'), Sales_Target=('Sales_Target','sum'), Customer_Satisfaction=('Customer_Satisfaction','mean'), On_Time_Delivery=('On_Time_Delivery','mean'))
summary['Sales_Achievement_%'] = summary['Sales_Actual'] / summary['Sales_Target'] * 100
display(summary.round(2))
```

## Analytical note

Observed relationships are associations, not proof of causation. Monthly performance data contains repeated employee-month observations, so it should not be interpreted as unique employee counts.
